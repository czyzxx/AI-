```python
# 折线图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
rcParams['font.family']='SimHei' #显示中文字体

# 创建图表，设置大小
plt.figure(figsize=(10,5))
month=['1月','2月','3月','4月']
sales=[100,150,80,130]

# 绘制折线图
plt.plot(month,sales,
         label='产品A',
         color='green',
         linestyle='--',
         linewidth=2,
         marker='o',)

# 添加标题
plt.title('2025销售趋势',color='yellow',fontsize=20)

# 添加坐标轴标签
plt.xlabel('月份',fontsize=10)
plt.ylabel('销售额（万元）',fontsize=10)

# 添加图例
plt.legend(loc='upper left')

# 添加网格线
plt.grid(True,alpha=0.3,color='white',linestyle='--') #横竖都有 alpha透明度
#plt.grid(axis='y') 只有横线
#plt.grid(axis='x') 只有竖线

# 设置刻度字体大小
plt.xticks(rotation=30,fontsize=12) #rotation字体旋转角度
plt.yticks(rotation=30,fontsize=12)

# 设置y轴范围
plt.ylim(60,160)

# 在每个数据点上显示数值
for x,y in zip(month,sales):
    print(x,y)
    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10) #ha:center居中 left居左 right居右  va:bottom底部对齐

# 显示图表
plt.show()
```

```python
# 柱状图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
rcParams['font.family']='SimHei' #显示中文字体

# 创建图表，设置大小
plt.figure(figsize=(10,5))
subject=['语文','数学','英语','科学']
scores=[85,92,78,88]

# 绘制柱状图
plt.bar(subject,scores,
         label='xxx',
         color='yellow',
         width=0.4)

# 添加标题
plt.title('xxx2025成绩',color='pink',fontsize=20)

# 添加坐标轴标签
plt.xlabel('科目',fontsize=10)
plt.ylabel('分数',fontsize=10)

# 添加图例
plt.legend(loc='upper left')

# 添加网格线
# plt.grid(True,alpha=0.3,color='white',linestyle='--') 横竖都有 alpha透明度
plt.grid(axis='y') # 只有横线
#plt.grid(axis='x') 只有竖线

# 设置刻度字体大小
plt.xticks(rotation=0,fontsize=12) #rotation字体旋转角度
plt.yticks(rotation=0,fontsize=12)

# 设置y轴范围
plt.ylim(0,100)

# 在每个数据点上显示数值
for x,y in zip(subject,scores):
    print(x,y)
    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10) #ha:center居中 left居左 right居右  va:bottom底部对齐

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 条形图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
rcParams['font.family']='SimHei' #显示中文字体

# 创建图表，设置大小
plt.figure(figsize=(10,5))
countries=['United States','China','Japan','Germany','India']
GDP=[85,92,43,32,8]

# 绘制条形图
plt.barh(countries,GDP)

# 添加标题
plt.title('2025GDP排名',color='yellow',fontsize=20)

# 添加坐标轴标签
plt.xlabel('GDP',fontsize=10)
plt.ylabel('国家',fontsize=10)

# 添加网格线
# plt.grid(True,alpha=0.3,color='white',linestyle='--') 横竖都有 alpha透明度
#plt.grid(axis='y')  只有横线
plt.grid(axis='x') # 只有竖线

# 设置刻度字体大小
plt.xticks(rotation=0,fontsize=12) #rotation字体旋转角度
plt.yticks(rotation=0,fontsize=12)

# 设置y轴范围
plt.xlim(0,100)

# 在每个数据点上显示数值
for x,y in zip(GDP,countries):
    print(x,y)
    plt.text(x+1,y,str(x),ha='center',va='bottom',fontsize=10) #ha:center居中 left居左 right居右  va:bottom底部对齐

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
rcParams['font.family']='SimHei' #显示中文字体

# 创建图表，设置大小
plt.figure(figsize=(10,5))
things=['学习','娱乐','运动','睡觉','其他']
times=[6,4,1,8,5]
colors=['#66b3ff','#99ff99','#ffcc99','#ff9999','#ff4499']

# 绘制饼图
plt.pie(times,labels=things,
        autopct='%1.1f%%',
        startangle=90, #调整初始画图的角度
        colors=colors,
        wedgeprops={'width':0.5}, # 圆环
        pctdistance=0.8 #调整百分比的位置
        )

# 添加标题
plt.title('一天的时间分布',color='white',fontsize=20)
plt.text(0,0,'总计：\n100%',ha='center',va='center',fontsize=10)

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 爆炸饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
rcParams['font.family']='SimHei' #显示中文字体

# 创建图表，设置大小
plt.figure(figsize=(10,5))
things=['学习','娱乐','运动','睡觉','其他']
times=[6,4,1,8,5]
colors=['#66b3ff','#99ff99','#ffcc99','#ff9999','#ff4499']
explode=[0.1,0,0,0,0] #设置突出块的位置

# 绘制饼图
plt.pie(times,labels=things,
        autopct='%1.1f%%',
        colors=colors,
        explode=explode, # 设置突出块
        shadow=True
        )

# 添加标题
plt.title('一天的时间分布',color='white',fontsize=20)

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 散点图
import matplotlib.pyplot as plt
import random
from matplotlib import rcParams # 字体
rcParams['font.family']='SimHei' #显示中文字体

# 创建图表，设置大小
plt.figure(figsize=(10,8))
x=[]
y=[]
for i in range(1000):
    tmp=random.uniform(0,10)
    x.append(tmp)
    tmp2=2*tmp+random.gauss(0,2) #2*tmp是理想直线y=2x         random.gauss(0,2)生成随机噪声，生成的数大部分在-2~2
    y.append(tmp2)

# 绘制散点图
plt.scatter(x,y,
            color='#66b3ff',
            alpha=0.5,
            s=20,
            label='数据')

# 添加标题
plt.title('x与y的关系',color='white',fontsize=20)

# 添加坐标轴标签
plt.xlabel('x自变量',fontsize=10)
plt.ylabel('y因变量',fontsize=10)

# 添加图例
plt.legend(loc='upper left')

# 添加网格线
plt.grid(True,alpha=0.3,color='white',linestyle='--') #横竖都有 alpha透明度
#plt.grid(axis='y') 只有横线
#plt.grid(axis='x') 只有竖线

# 设置刻度字体大小
plt.xticks(rotation=0,fontsize=12) #rotation字体旋转角度
plt.yticks(rotation=0,fontsize=12)

# 设置y轴范围
plt.ylim(0,25)

#画折线图
plt.plot([0,10],[0,20],color='#99ff99',linestyle='-',linewidth=2)

# 显示图表
plt.show()
```

```python
#箱线图
import matplotlib.pyplot as plt

data={
    '语文':[82,85,88,70,90,76,84,83,95],
    '数学':[75,80,79,93,88,82,87,89,92],
    '英语':[70,72,68,65,78,80,85,90,95]
}
plt.figure(figsize=(8,6))
plt.boxplot(data.values(),tick_labels=data.keys())

plt.title("各科成绩分布（箱线图）")
plt.ylabel("分数")
plt.grid(True,axis='y',linestyle='--',alpha=0.5)
plt.show()
```

箱线图（boxplot）各部分含义

1.最上面横线（上须）：正常数据里的最大值
2.箱子上边：上四分位数 Q3 → 75%的数据 ≤这个分数
3.箱子中间橙色横线：中位数，一半人高于这个分，一半低于
4.箱子下边：下四分位数 Q1 → 25%的数据 ≤这个分数
5.最下面横线（下须）：正常数据里的最小值
6.圆圈/圆点（语文左下角那个）：异常值（离群点），分数明显偏离大部分同学

通俗记

- 箱子：中间50%大部分人的成绩，箱子越宽，这科中间同学分数差距越大
- 中间橙线：中等水平
- 须线：除去异常值后的最高、最低
- 小圆点：考得特别离谱（过低或过高）的个别同学

举例：
英语箱子很长 → 英语大家分数差距很大；
语文出现圆点 → 有一个同学分数明显偏低，是异常样本。
