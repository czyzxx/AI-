## 缺失值的处理

```python
# nan:not a number
import pandas as pd
import numpy as np
s=pd.Series([1,2,np.nan,None,pd.NA])
df=pd.DataFrame([[1,pd.NA,2],[2,3,5],[None,4,6]],columns=['A','B','C'])
print(s)
# 查看是否是缺失值
print(s.isna())
print(s.isnull())
print(df.isna())
print(df.isnull())
print(df.isna().sum()) # 查看缺失值的个数
print(s.isnull().sum())
print('-'*30)

#剔除缺失值
print(s.dropna())
print('-'*30)
print(df.dropna())  #直接把一行删除
print('-'*30)
print(df.dropna(how='all'))  #全部是缺失值才删除
print('-'*30)
print(df.dropna(thresh=2)) # 如果至少有n个值不是缺失值就保留
print('-'*30)
print(df.dropna(axis=1)) # 剔除一整列的记录
print('-'*30)
print(df.dropna(subset=['A'])) #如果某列有缺失值，则删除这一行

#填充缺失值
df=pd.read_csv('titanic.csv')
df.tail()
df.isna().sum(axis=0)
df.head()
print(df.fillna({'Age':0}).tail()) #使用字典来填充
print('-'*30)
print(df.fillna(df[['Age']].mean()).tail()) #使用统计值来填充缺失值
print('-'*30)
print(df.ffill().tail()) #front
print('-'*30)
print(df.bfill().tail()) #behind
```

## 数据变形

```python
import pandas as pd
data={
    'ID':[1,2],
    'name':['A','B'],
    'Math':[90,85],
    'English':[88,92],
    'Science':[95,89]
}
df=pd.DataFrame(data)
print(df)
print('-'*30)
print(df.T) #行列转置
```

```python
#宽表转换成长表
df2=pd.melt(df,id_vars=['ID','name'],var_name='subject',value_name='scores')
df2.sort_values('name')
print(df2)
```

```python
#长表转宽表
df3=pd.pivot(df2,index=['ID','name'],columns='subject',values='scores')
print(df3)
```

```python
data={
    'ID':[1,2],
    'name':['A C','B C'],
    'Math':[90,85],
    'English':[88,92],
    'Science':[95,89]
}
df=pd.DataFrame(data)
#分列
df[['first','last']]=df['name'].str.split(" ",expand=True)
print(df)
```

```python
data = {
    "person_id": [1,2,3,4],
    "blood_pressure": ["120/80","130/85","115/75","142/90"]
}
df = pd.DataFrame(data)
print(df)
print('-'*30)
df = df[['person_id','blood_pressure']]
df[['high','low']] = df['blood_pressure'].str.split("/", expand=True)
print(df)
```

## 时间数据的处理

```python
import pandas as pd
d=pd.Timestamp('2015-05-02 10:22')
d1=pd.Timestamp('2015-05-02 13:22')
print(d)
print(type(d))
print("年",d.year)
print("月",d.month)
print("日",d.day)
print(d.hour,d.minute,d.second)
print("季度",d.quarter)
print("是否是月底",d.is_month_end)
print("星期",d.day_name())
print("转化为天",d.to_period("D"))
print("转化为天",d1.to_period("D"))
print("转化为季度",d.to_period("Q"))
print("转化为年",d.to_period("Y"))
print("转化为月",d.to_period("M"))
print("转化为周",d.to_period("W"))
```

```python
# 字符串转换为日期类型
a=pd.to_datetime('2015-02-28 10:22')
print(a)
print(type(a))
print(a.day_name())
```

```python
# dataFrame日期转换
df=pd.DataFrame({
    'sales':[100,200,300],
    'date':['20250601','20250602','20250603']
})
df['datetime']=pd.to_datetime(df['date'])
print(df)
print('-'*30)
print(df.info()) #查看表格整体信息，看每列数据类型、有无空值
print('-'*30)
print(type(df['datetime']))
df['week']=df['datetime'].dt.day_name() #因为类型为Series，所以要加dt.进行转换
print(df['week'])
print('-'*30)
print(df['datetime'].dt.year)
```

## 分组聚合

```python
#df.groupby('分组的字段')['聚合的字段'].聚合函数()
import pandas as pd
df=pd.read_csv('titanic.csv')
df['Pclass']=df['Pclass'].astype('int64')
#计算不同Pclass的评价Fare
df.groupby('Pclass').groups #查看分组
```

```python
print(df.groupby('Pclass').get_group(1)) #查看具体的某个分组数据
```

```python
df.groupby('Pclass')['Fare'].mean()
```

单层['列名']-->Series
双层[['列名']]-->DataFrame

```python
df2=df.groupby('Pclass')[['Fare']].mean()
df2['Fare']=df2['Fare'].round(2) #round(2)保留两位小数点
print(df2)
```

```python
#df2=df2.reset_index()
print(df2.sort_values('Fare',ascending=False)) #从高到低
```

```python
# 根据多个条件分组
df2=df.groupby(['Pclass','Sex'])[['Fare']].mean()
df2=df2.reset_index()
df2['Fare']=df2['Fare'].round(2)
print(df2)
```
